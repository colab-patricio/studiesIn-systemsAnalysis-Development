### O que são variáveis?
Uma variável é um espaço reservado na memória do computador para armazenar dados que podem ser modificados e acessados ao longo da execução de um programa.

#### A. Variáveis inteiras - int

Usadas para armazenar números inteiros, sem casas decimais. Perfeitas para contagens, idades, quantidades, etc.

```java
int idade = 30;
int quantidadeDeProdutos = 10;
```

#### B. Variáveis do tipo flutuante - float

Utilizadas para números que possuem casas decimais. Essenciais para preços, médias, porcentagens, altura, peso, etc.
```java
float altura = 1.75f;
```

#### C. Variáveis de texto (string) - str
Usadas para armazenar sequências de caracteres, ou seja, texto. São definidas com aspas duplas ou simples, dependendo da linguagem
```java
String nomeCompleto = "João Silva Santos";
String mensagemDeErro = "Login inválido.";
```

#### Variáveis booleanas (boolean) - bool
Armazenam apenas dois valores possíveis: verdadeiro (`true`) ou falso (`false`). São a base para toda a tomada de decisão em um programa.

```java
boolean usuarioEstaLogado = true;
boolean pagamentoAprovado = false;
```


#### Regras Universais para nomes

Apesar dessas diferenças, quase todas as linguagens concordam em um conjunto de regras básicas para os nomes (as "etiquetas"):

- Não pode ter espaços: É por isso que nome do usuario não funciona. Para resolver isso, usamos padrões como nomeDoUsuario (o camelCase) ou nome_do_usuario (o snake_case).
- Não pode começar com números: Um nome como 1ano vai dar erro, mas ano1 é perfeitamente válido.
- Não pode usar "palavras reservadas": Cada linguagem tem palavras especiais que usa para seus próprios comandos (como if, for, while). Você não pode usar essas palavras como nomes de variáveis.

> Nota: snake_case (Caixa de Cobra): É o padrão mais usado em Python. Aqui, todas as palavras são minúsculas e separadas por um underline ( _ ), como em **nome_do_usuario**.

### Aula 07 - Operadores aritméticos
Nessa aula iremos aprender sobre operadores aritméticos e sua ordem de procedência dentro de expressões matemáticas em Python.

| Operador aritmético | Operação |
|---------------------|----------|
| + | Adição |
| - | Subtração |
| * | Multiplicação |
| / | Divisão |
| ** | Potência |
| // | Divisão inteira |
| % | Módulo (resto de uma divisão) |

> Nota: Ordem de precedência: I. (); II. **; III. *, /, //, %; & IV. + ou -


### Exemplos

````python
num1 = int(input('Digite um valor inteiro: '))
num2 = int(input('Digite outro valor inteiro: '))

soma = num1 + num2
multiplicacao = num1 * num2
div = num1 / num2
exp = num1 ** num2
div_inteira = num1 // num2
mod = num1 % num2

print('Números digitados: {} e {}'.format(num1, num2))
print('A soma entre os números digitados é igual a {} e o produto é igual a {}.'.format(soma, multiplicacao))
print('A divisão é igual a {:.2f} enquanto a potência é igual a {}.'.format(div, exp))
print('A divisão inteira é igual a {} e o modulo igual a {}'.format(div_inteira, mod))
````
